#1. 단변량 분석 (Univariate Analysis) - 변수 1개
(1) 수치형 변수 (total_bill, tip)

히스토그램 (Histplot / KDE): 값의 분포와 왜도(Skewness) 파악

박스플롯 (Boxplot): 중앙값, 사분위수 및 이상치(Outlier) 확인

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 테마 및 스타일 설정
sns.set_theme(style="whitegrid")

# tips 데이터셋 로드
tips = sns.load_dataset("tips")
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 1. Total Bill 히스토그램 + 밀도 곡선(KDE)
sns.histplot(tips['total_bill'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Total Bill Distribution')

# 2. Tip 박스플롯
sns.boxplot(y=tips['tip'], ax=axes[1], color='lightgreen')
axes[1].set_title('Tip Boxplot (Outliers Check)')

plt.tight_layout()
plt.show()
```

(2) 범주형 변수 (day, sex, time, smoker)

카운트플롯 (Countplot): 각 범주별 데이터 개수(빈도) 비교

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 1. 요일별 방문 빈도
sns.countplot(x='day', data=tips, ax=axes[0], palette='Set2')
axes[0].set_title('Count by Day')

# 2. 식사 시간대별 방문 빈도
sns.countplot(x='time', data=tips, ax=axes[1], palette='Set2')
axes[1].set_title('Count by Time')

plt.tight_layout()
plt.show()
```

#2. 이변량 분석 (Bivariate Analysis) - 변수 2개
(1) 수치형 vs 수치형 (total_bill vs tip)

산점도 + 회귀선 (Regplot): 두 양적 변수 간의 선형 상관관계 파악

```python
plt.figure(figsize=(7, 4))
sns.regplot(x='total_bill', y='tip', data=tips, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Total Bill vs Tip (Correlation)')
plt.show()
```

(2) 범주형 vs 수치형 (day vs total_bill)

박스플롯 / 바플롯: 그룹별 분포 차이 및 평균 비교


```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

# x축에 수치형(total_bill), y축에 범주형(day)을 지정하여 가로 박스플롯 생성
sns.boxplot(
    x='total_bill',
    y='day',
    data=tips,
    palette='Pastel1'
)

plt.title('Total Bill by Day', fontsize=14)
plt.xlabel('Total Bill ($)', fontsize=12)
plt.ylabel('Day of Week', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns

# 1. tips 데이터셋 로드
tips = sns.load_dataset("tips")

# 2. 요일(day)별 팁 분석 (총합, 평균, 최대값, 건수)
day_tip = tips.groupby('day', observed=False)['tip'].agg(['sum', 'mean', 'max', 'count']).reset_index()
day_tip.columns = ['요일', '팁 총합($)', '팁 평균($)', '최대 팁($)', '결제 건수']

# 3. 식사 시간대(time)별 팁 분석
time_tip = tips.groupby('time', observed=False)['tip'].agg(['sum', 'mean', 'max', 'count']).reset_index()
time_tip.columns = ['시간대', '팁 총합($)', '팁 평균($)', '최대 팁($)', '결제 건수']

# 4. 요일 & 시간대 복합 그룹별 팁 분석
combined_tip = tips.groupby(['day', 'time'], observed=False)['tip'].agg(['sum', 'mean', 'max', 'count']).dropna().reset_index()
combined_tip.columns = ['요일', '시간대', '팁 총합($)', '팁 평균($)', '최대 팁($)', '결제 건수']

print("=== 1. 요일별 팁 집계 ===")
display(day_tip.sort_values(by='팁 총합($)', ascending=False))

print("\n=== 2. 식사 시간대별 팁 집계 ===")
display(time_tip.sort_values(by='팁 총합($)', ascending=False))

print("\n=== 3. 요일 + 시간대 조합별 팁 집계 ===")
display(combined_tip.sort_values(by='팁 총합($)', ascending=False))
```

```python
# 가장 팁 총합이 많은 요일과 시간대 찾기
max_day_sum = day_tip.loc[day_tip['팁 총합($)'].idxmax()]
max_time_sum = time_tip.loc[time_tip['팁 총합($)'].idxmax()]

# 단일 결제 기준 가장 큰 팁을 준 손님 정보
max_single_tip = tips.loc[tips['tip'].idxmax()]

print(f"📌 가장 많은 팁 총액을 기록한 요일: {max_day_sum['요일']} (${max_day_sum['팁 총합($)']:.2f})")
print(f"📌 가장 많은 팁 총액을 기록한 시간대: {max_time_sum['시간대']} (${max_time_sum['팁 총합($)']:.2f})")
print(f"📌 단일 결제 기준 최고 팁: ${max_single_tip['tip']:.2f} ({max_single_tip['day']} {max_single_tip['time']}, {max_single_tip['sex']})")
```

```python
# 요일별(col), 시간대별(row), 성별(hue)로 분할하여 전체 패턴 관찰

# 1. Seaborn 테마를 어두운 배경(darkgrid)으로 설정합니다.
sns.set_theme(style="darkgrid")

sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',          # 색상: 성별
    style='smoker',     # 마커 모양: 흡연 여부
    col='day',          # 열 분할: 요일
    row='time',         # 행 분할: 시간대 (Lunch / Dinner)
    height=2.5,
    aspect=1.2
)
plt.show()
```

### 그려서 median_house_value/population 별로 scatter하여 어느지역의 집값이 비싼지 말해줘? 시각화도 해줘...

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.datasets import fetch_california_housing

# 1. 캘리포니아 주택 데이터셋 로드
california = fetch_california_housing(as_frame=True)
df = california.frame

# 2. 인구 대비 집값 변수 생성
df['value_per_pop'] = (df['MedHouseVal'] * 100000) / df['Population']

# [개선] 테마 설정 (ticks)
sns.set_theme(style="ticks")

plt.figure(figsize=(10, 8))

# 3. 데이터 시각화 (합쳐서 그리기)
# x, y: 위치 (경도, 위도)
# c: 색상 (Median House Value)
# s: 크기 (Value per Population) - 비율을 강조하기 위해 스케일을 조정했습니다.
scatter = plt.scatter(
    x=df['Longitude'],
    y=df['Latitude'],
    c=df['MedHouseVal'],    # 색상은 집값으로
    cmap='coolwarm',         # coolwarm 컬러맵 사용
    alpha=0.5,              # 투명도
    s=df['value_per_pop'] * 0.01, # 점 크기는 인구 대비 집값 비율로 (스케일 조정)
    label='Median House Value',
)

# [개선] 크기 범례 추가
# value_per_pop의 대표적인 값들을 사용하여 크기 범례를 만듭니다.
pop_ratios = [df['value_per_pop'].quantile(0.25), df['value_per_pop'].median(), df['value_per_pop'].quantile(0.75)]
for ratio in pop_ratios:
    plt.scatter([], [], c='black', alpha=0.3, s=ratio * 0.01, label=f'Pop. Ratio: {ratio:.0f}')

# 그래프 정보 추가
plt.title('California House Value vs. Population Ratio', fontsize=16)
plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.legend(title='Metrics', fontsize=10, bbox_to_anchor=(1.05, 1), loc='upper left') # 범례 추가 및 위치 조정
plt.colorbar(scatter, label='Median House Value ($100k)') # 색상바 추가

# 레이아웃 조정
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. tips 데이터셋 로드
tips = sns.load_dataset("tips")

# 2. 요일별 집계 (팁 총합, 팁 평균, 방문 건수, 총 방문 인원수)
day_summary = (
    tips.groupby('day', observed=False)
    .agg(
        tip_sum=('tip', 'sum'),
        tip_mean=('tip', 'mean'),
        count=('tip', 'count'),  # 결제 건수(팀 수)
        size_sum=('size', 'sum'),  # 총 방문 인원수
    )
    .reset_index()
)

# 3. 요일 이름 한글 변경 및 순서 정렬
day_map = {'Thur': '목요일', 'Fri': '금요일', 'Sat': '토요일', 'Sun': '일요일'}
day_summary['day_kr'] = day_summary['day'].map(day_map)

# 결과 테이블 출력
print("=== 요일별 팁 및 방문객 수 요약 ===")
display(
    day_summary[
        ['day_kr', 'count', 'size_sum', 'tip_sum', 'tip_mean']
    ].rename(
        columns={
            'day_kr': '요일',
            'count': '결제 건수(팀)',
            'size_sum': '총 인원수(명)',
            'tip_sum': '팁 총액($)',
            'tip_mean': '평균 팁($)',
        }
    )
)

# 4. 가장 많은 요일 자동 추출
top_tip_day = day_summary.loc[day_summary['tip_sum'].idxmax()]
top_people_day = day_summary.loc[day_summary['size_sum'].idxmax()]
top_avg_tip_day = day_summary.loc[day_summary['tip_mean'].idxmax()]

print(
    f"📌 팁이 가장 많은 요일 (총액 기준): {day_map[top_tip_day['day']]} (${top_tip_day['tip_sum']:.2f})"
)
print(
    f"📌 사람이 가장 많은 요일 (총 인원 기준): {day_map[top_people_day['day']]} ({top_people_day['size_sum']}명, {top_people_day['count']}팀)"
)
print(
    f"📌 팁을 가장 많이 주는 요일 (평균 기준): {day_map[top_avg_tip_day['day']]} (${top_avg_tip_day['tip_mean']:.2f})"
)

# 5. 한눈에 보는 시각화
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# (1) 요일별 총 방문 인원수 (사람이 가장 많은 요일)
sns.barplot(
    data=day_summary,
    x='day_kr',
    y='size_sum',
    ax=axes[0],
    palette='Blues_d',
    hue='day_kr',
    legend=False,
)
axes[0].set_title('Total Visitors by Day (People Count)', fontsize=13)
axes[0].set_xlabel('Day of Week')
axes[0].set_ylabel('Total People')

# 막대 위에 숫자 표시
for p in axes[0].patches:
    axes[0].annotate(
        f'{int(p.get_height())}명',
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha='center',
        va='center',
        xytext=(0, 5),
        textcoords='offset points',
    )

# (2) 요일별 총 팁 금액 (팁이 가장 많은 요일)
sns.barplot(
    data=day_summary,
    x='day_kr',
    y='tip_sum',
    ax=axes[1],
    palette='Greens_d',
    hue='day_kr',
    legend=False,
)
axes[1].set_title('Total Tip Amount by Day ($)', fontsize=13)
axes[1].set_xlabel('Day of Week')
axes[1].set_ylabel('Total Tip ($)')

# 막대 위에 금액 표시
for p in axes[1].patches:
    axes[1].annotate(
        f'${p.get_height():.1f}',
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha='center',
        va='center',
        xytext=(0, 5),
        textcoords='offset points',
    )

plt.tight_layout()
plt.show()
```
